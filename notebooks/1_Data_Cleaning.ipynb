```python
# Visualizing Missing Values

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.heatmap(df.isnull(), cbar=False)
plt.show()
```

Handling Missing Values

```python
## AGE

sns.set_palette('coolwarm')
sns.histplot(df['age'],bins=15, kde=True)
plt.xlabel("Patient's Age")
plt.ylabel("Number of Patients")
plt.title('Age Distribution')
plt.show()

print(f"Age Distribution Skewness: {df['age'].skew()}")
```

```python
df['age'] = df['age'].fillna(df['age'].median())
```

```python
## AVERAGE TEMPERATURE PER DAY

sns.histplot(df['average_temp_day'],bins=15,kde=True)
plt.title('Avg Temperature/day Distribution')
plt.show()

print(f"Avg temperature skewness: {df['average_temp_day'].skew()}")
```

```python
df['average_temp_day'] = df['average_temp_day'].fillna(df['average_temp_day'].mean())
```

```python
## MAXIMUM TEMPERATURE PER DAY

sns.histplot(df['max_temp_day'],bins=15,kde=True)
plt.title('Max Temperature/day Distribution')
plt.show()

print(f"Max temperature skewness: {df['max_temp_day'].skew()}")
```

```python
df['max_temp_day'] = df['max_temp_day'].fillna(df['max_temp_day'].mean())
```

```python
## AVERAGE RAIN PER DAY

sns.histplot(df['average_rain_day'],bins=15,kde=True)
plt.title('Avg Rain/day Distribution')
plt.show()

print(f"Avg Rain skewness: {df['average_rain_day'].skew()}")
```

```python
df['average_rain_day'] = df['average_rain_day'].fillna(df['average_rain_day'].median())
```

```python
## MAX RAIN PER DAY

sns.histplot(df['max_rain_day'],bins=15,kde=True)
plt.title('Max Rain/day Distribution')
plt.show()

print(f"Max Rain skewness: {df['max_rain_day'].skew()}")
```

```python
df['max_rain_day'] = df['max_rain_day'].fillna(df['max_rain_day'].median())
```

```python
# Categorical columns

## speciality, disability, place

df['specialty'] = df['specialty'].fillna('unknown')  # missing percentage - 18.37
df['place'] = df['place'].fillna('unknown')  # missing percentage - 15.15
df['disability'] = df['disability'].fillna('unknown')  # missing percentage - 10.53
```
